```python
trend_results = []

for province in df["province"].unique():

    temp = df[df["province"] == province]

    annual = temp.resample("YE").mean(numeric_only=True)

    x = np.arange(len(annual))

    slope, intercept, r, p, stderr = linregress(
        x,
        annual["daily_rainfall"]
    )

    trend_results.append({
        "Province": province,
        "Slope": slope,
        "P-value": p
    })

trend_df = pd.DataFrame(trend_results)

print(trend_df)
```

```python
import pandas as pd
import os

# Load the dataset
df = pd.read_csv("niwis_daily_climate_master.csv")

# Create an output folder
output_folder = "South_Africa_Provinces"
os.makedirs(output_folder, exist_ok=True)

# Convert date column (optional but recommended)
df["date"] = pd.to_datetime(df["date"])

# Sort the data
df = df.sort_values(["province", "date"])

# Split into individual province files
for province, data in df.groupby("province"):
    
    # Create a safe filename
    filename = province.replace(" ", "_").replace("-", "_")
    
    # Save the province data
    data.to_csv(
        os.path.join(output_folder, f"{filename}.csv"),
        index=False
    )

    print(f"Saved: {filename}.csv ({len(data)} rows)")

print("\nAll province datasets have been created successfully!")
```
